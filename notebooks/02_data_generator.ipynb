## Genrerating realistic fake database using `FAKER` librabry

```python
import pandas as pd 
import numpy as np
from faker import Faker
import random
from datetime import datetime, timedelta

fake = Faker('en_IN') # Use Indian locale for realistic data
```

```python
mumbai_areas = ['Andheri', 'Bandra', 'Borivali', 'Dadar', 'Ghatkopar', 'Juhu', 'Kurla', 'Malad', 'Mulund', 'Powai', 'Santacruz', 'Vile Parle']
car_types = ['Sedan', 'SUV', 'Hatchback']
fuel_types = ['Petrol', 'Diesel', 'Electric', 'Hybrid']
ac_types = ["Ac", "Non-Ac"]
ride_status = ["Completed", "Completed", "Completed", "Completed", "Cancelled"] # 80% completed, 20% cancelled
```

```python
fare_rates = {
  "Sedan" : 15,
  "SUV" : 19,
  "Hatchback" : 11
}
base_fare = 50
```

```python
end_date = datetime.now()
start_date = end_date - timedelta(days=365) # Last 1 year

print(f"Start Date: {start_date}")
print(f"End Date: {end_date}")
```

```python
import pandas as pd 
import time

start_time = time.time()
ride_booking = []

def data_generator():
    car_type = random.choice(car_types)
    distance_km = round(random.uniform(1, 30), 2) # Distance between 1 and 30 km
    Total_Fare = base_fare + fare_rates[car_type] * distance_km # Base fare + (rate per km * distance)
    ride_booking.append({
        "ride_id": fake.uuid4(),
        "driver_name": fake.name(),
        "driver_phone": fake.phone_number(),
        "driver_rating": round(random.uniform(3.5, 5.0), 1), # Rating between 3.5 and 5.0
        "driver_experience_years": random.randint(1, 20), # Experience between 1 and 20 years
        "driver_vehicle_number": fake.license_plate(),
        "driver_vehicle_model": fake.vehicle_make_model(),
        "customer_name": fake.name(),
        "customer_phone": fake.phone_number(),  
        "pickup_location": random.choice(mumbai_areas),
        "drop_location": random.choice(mumbai_areas),
        "car_type": car_type,
        "fuel_type": random.choice(fuel_types),
        "ac_type": random.choice(ac_types),
        "ride_status": random.choice(ride_status),
        "distance_km": distance_km, 
        "fare": Total_Fare, 
        "booking_time": fake.date_time_between(start_date=start_date, end_date=end_date)
    })

for i in range (100000):
    data_generator()

end_time = time.time()
print(f"Data generation completed in {end_time - start_time:.2f} seconds")

df = pd.DataFrame(ride_booking)
print(df.head())

```
