## Generating and storing unique Licence plate numbeer

```python
import pandas as pd 
import random
import string 
import time 

start_rand_plate_num = time.time()
used_numbers = set()       #  set  → validation / lookup (fast checking)
plates = []                   # list → storage / ordering (ordered storage)

def license_plate_generator():
    state = 'MH' # Maharashtra state code for license plates
    district_code = random.randint(1, 50) # Random district code between 1 and 50

    letters = string.ascii_uppercase # Uppercase letters for license plate
    char = ''.join(random.choices(letters, k=2)) # Random 2 letters for license plate

    numeric_part = f"{random.randint(0, 9999):04d}" # Random 4 digit number for license plate, formated 04d means fill empty space with 0 and generate 4 digit number d means integer
    return f"{state}{district_code} {char} {numeric_part}"

end_rand_plate_num = time.time()

start_unique_plate = time.time()

def generate_unique_license_plate():
    while True:
        license_plate = license_plate_generator()
        if license_plate not in used_numbers:
            used_numbers.add(license_plate)
            plates.append(license_plate)        # just to create a list to check

            return license_plate

for i in range(10000):
        vehicle_license_plate = generate_unique_license_plate()
        
end_unique_plate = time.time()

print(f"Random plate generated : {end_rand_plate_num - start_rand_plate_num:.2}") 
time.sleep(1)
print(f"Unique plate identified and stored: {end_unique_plate - start_unique_plate:.2}\n")
time.sleep(1)

print(f"Generated License Plate: {vehicle_license_plate}")
time.sleep(1)
print(f"Unique plate numbers: {(plates)[-5:]}")        # last 5 plate numbers
time.sleep(1)
print(f"Total num of unique plates generated: {len(used_numbers)} ")
time.sleep(1)

df = pd.DataFrame
```

# idempotency
Running the same code multiple times should not corrupt or change existing valid data.

```python
import pandas as pd 
import random 
from faker import Faker 

faker = Faker('en_IN')  

# idempotency
import os

if not os.path.exists("../data/drivers.csv"):   # Check if file exists if yes True ---> Fale (reverse result)
    print("Generating new drivers...")
    
    driver_info = []
    letters = 'DRV'

    def drivers_data(driver_num):
        vehical_model = random.choice(list(vehicle_category.keys()))
        driver_info.append({
            "driver_id" : f"{letters}{driver_num+1:04d}",
            "driver_name" : faker.name(),
            "driver_phone": faker.phone_number(),
            "license_plate" : plates[driver_num],
            "driver_vehicle_model": vehical_model,  
            "car_type": vehicle_category[vehical_model],        # dictonary lookup
            "driver_rating": round(random.uniform(3.5, 5.0), 1), # Rating between 3.5 and 5.0
            "driver_experience_years": random.randint(1, 20), # Experience between 1 and 20 years
        })

    for driver_num in range(10000):
        drivers_data(driver_num)

    df = pd.DataFrame(driver_info)
    df.index = range(1,len(df) + 1)

    df.to_csv("../data/drivers.csv" , sep = ",", index=False )

else:
    print("Drivers already exist")
    df = pd.read_csv("../data/drivers.csv")
df
```
